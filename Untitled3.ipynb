```python
from insightface.app import FaceAnalysis

app = FaceAnalysis(name="buffalo_l")
app.prepare(ctx_id=-1)

print("Yüz tanıma modeli hazır.")
```

```python
os.makedirs("faces/goktug", exist_ok=True)
os.makedirs("faces/test/goktug", exist_ok=True)
os.makedirs("faces/test/arkadas", exist_ok=True)

print("Proje klasörleri hazır.")
```

```python
reference_images = os.listdir("faces/goktug")

print("Referans fotoğraf sayısı:", len(reference_images))
print("Dosyalar:")

for filename in reference_images:
    print("-", filename)
```

```python
for filename in reference_images:
    image_path = os.path.join("faces/goktug", filename)

    image = cv2.imread(image_path)
    faces = app.get(image)

    print(f"{filename} → {len(faces)} yüz bulundu")
```

```python
goktug_embeddings = []

for filename in reference_images:
    image_path = os.path.join("faces/goktug", filename)

    image = cv2.imread(image_path)
    faces = app.get(image)

    if len(faces) == 1:
        goktug_embeddings.append(faces[0].embedding)

print("Oluşturulan yüz temsili:", len(goktug_embeddings))
print("Temsil boyutu:", goktug_embeddings[0].shape)
```

```python
goktug_embedding = np.mean(goktug_embeddings, axis=0)

print("Göktuğ yüz profili oluşturuldu.")
print("Profil boyutu:", goktug_embedding.shape)
```

```python
print("Göktuğ test fotoğraflarını yükle:")
uploaded = files.upload()

for filename in uploaded.keys():
    shutil.move(filename, os.path.join("faces/test/goktug", filename))

print("Göktuğ test fotoğrafları hazır.")
```

```python
print("Arkadaş test fotoğraflarını yükle:")
uploaded = files.upload()

for filename in uploaded.keys():
    shutil.move(filename, os.path.join("faces/test/arkadas", filename))

print("Arkadaş test fotoğrafları hazır.")
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))


for person in ["goktug", "arkadas"]:
    folder_path = os.path.join("faces/test", person)

    print(f"\n--- {person.upper()} TESTLERİ ---")

    for filename in sorted(os.listdir(folder_path)):
        image_path = os.path.join(folder_path, filename)

        image = cv2.imread(image_path)
        faces = app.get(image)

        if len(faces) != 1:
            print(f"{filename} → Yüz sayısı: {len(faces)}")
            continue

        test_embedding = faces[0].embedding

        similarity = cosine_similarity(
            goktug_embedding,
            test_embedding
        )

        print(f"{filename} → Benzerlik: {similarity:.4f}")
```

```python
THRESHOLD = 0.50

results = []

for person in ["goktug", "arkadas"]:
    folder_path = os.path.join("faces/test", person)

    for filename in sorted(os.listdir(folder_path)):
        image_path = os.path.join(folder_path, filename)

        image = cv2.imread(image_path)
        faces = app.get(image)

        if len(faces) != 1:
            results.append({
                "Gerçek Kişi": person,
                "Dosya": filename,
                "Benzerlik": None,
                "Tahmin": "Yüz tespit edilemedi"
            })
            continue

        test_embedding = faces[0].embedding

        similarity = cosine_similarity(
            goktug_embedding,
            test_embedding
        )

        prediction = (
            "Göktuğ"
            if similarity >= THRESHOLD
            else "Tanınmayan kişi"
        )

        results.append({
            "Gerçek Kişi": person,
            "Dosya": filename,
            "Benzerlik": round(float(similarity), 4),
            "Tahmin": prediction
        })

for result in results:
    print(result)
```

```python
import pandas as pd

results_df = pd.DataFrame(results)

display(results_df)
```

```python
correct = 0
total = 0

for result in results:
    if result["Benzerlik"] is None:
        continue

    total += 1

    actual_person = result["Gerçek Kişi"]
    prediction = result["Tahmin"]

    if actual_person == "goktug" and prediction == "Göktuğ":
        correct += 1
    elif actual_person == "arkadas" and prediction == "Tanınmayan kişi":
        correct += 1

accuracy = (correct / total) * 100

print(f"Doğru sonuç: {correct}/{total}")
print(f"Doğruluk oranı: %{accuracy:.2f}")
```

# Göktuğ Face Recognition System

Bu proje, önceden eğitilmiş InsightFace modeli kullanılarak yüz tanıma işlemi gerçekleştirmektedir.

## Projenin Amacı

Sistem, referans fotoğraflarından Göktuğ'a ait bir yüz profili oluşturur ve test fotoğraflarındaki yüzleri bu profil ile karşılaştırır.

Yüzler arasındaki benzerlik, cosine similarity yöntemi kullanılarak hesaplanır.

## Kullanılan Teknolojiler

- Python
- OpenCV
- NumPy
- Pandas
- InsightFace
- Google Colab

## Sistem Akışı

1. Gerekli kütüphaneler yüklenir.
2. InsightFace yüz modeli hazırlanır.
3. Referans fotoğraflarından Göktuğ yüz profili oluşturulur.
4. Test fotoğrafları otomatik olarak okunur.
5. Yüz embedding'leri oluşturulur.
6. Cosine similarity hesaplanır.
7. Belirlenen threshold değerine göre tanıma yapılır.
8. Sonuçlar tablo halinde gösterilir.
9. Sistem doğruluk oranını hesaplar.
